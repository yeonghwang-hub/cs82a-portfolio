```python
import pandas as pd
import seaborn as sns
import scipy
import matplotlib.pyplot as plt

df = pd.read_csv("custdata.tsv", sep="\t")
df
```

```python
df_cleaned = df.dropna(subset=["housing.type"])
```

```python
df_cleaned
```

```python
sns.countplot(data=df_cleaned, x="housing.type")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```

```python
sorted_order = df["housing.type"].value_counts(ascending=True).index
sns.countplot(data=df_cleaned, x="housing.type", order=sorted_order)
plt.xticks(rotation=45, ha='right')
plt.title("Most People Own a home with Mortgage/Loan")
plt.xlabel("Housing Type")
plt.ylabel("Number of People")
plt.tight_layout()
```

From the original df with 1000 rows, I removed 56 rows with value "NaN" as housing type, leaving a clean df with 944 rows.
