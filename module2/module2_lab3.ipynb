```python
temperatures = [10, 16, 23, 28, 14, 22, 30]
```

"Freezing" if the temperature is below 12C

“Cold” if the temperature is between 12C and 15C (inclusive)

"Mild" if the temperature is between 16C and 20C (inclusive)

“Warm” if the temperature is between 21C and 25C (inclusive)

“Hot” if the temperature is above 25C

Write a for-loop to go through the list and label each temperature accordingly.
I expect temp 16C to print "Mild" since it is on the boundary between 16 and 20.

```python
for temp in temperatures:
    if temp < 12:
        print("Freezing")
    elif 12 <= temp <= 15:
        print("Cold")
    elif 16 <= temp <= 20:
        print("Mild")
    elif 21 <= temp <= 25:
        print("Warm")
    elif temp > 25:
        print("Hot")
```
