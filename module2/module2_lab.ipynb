```python
name = "Yeong"
hourly_wage = 30.5
hours = 8
```

compute weekly wage by multiplying hourly_wage and hours, and multiplying the result to 5 (# of work days per week)

```python
weekly_pay = (hourly_wage * hours) * 5
print(f"{name} earns {weekly_pay} per week.")
```

```python
temps = [61, 65, 72, 58, 70, 74, 68]
above_sixtyfive = 0
for t in temps:
    if t > 65:
        above_sixtyfive = above_sixtyfive + 1

print(above_sixtyfive)
```
