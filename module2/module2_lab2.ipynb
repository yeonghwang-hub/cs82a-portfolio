 Write a function with a for-loop

```python
def mult_for_loop(num1, num2):
    result = 0
    for i in range(num2):
        result = result + num1
    return result
```

```python
mult_for_loop(7, 6)
```

```python
mult_for_loop(3,9)
```

Write a function with a while loop

```python
def mult_while_loop(num1, num2):
    result = 0
    counter = 0
    while (counter < num2):
        result = result + num1
        counter = counter + 1
    return result
```

The while loop terminates once the variable "counter" becomes equal to num2 (the second argument). For example, counter starts with 0 and when given 7 and 6 as arguments into the function, the while loop runs 6 times, from counter being 0 all the way up to 5.

```python
mult_while_loop(7, 6)
```

```python
mult_while_loop(8,2)
```
