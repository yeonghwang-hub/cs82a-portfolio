```python
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import pandas as pd

df = pd.read_csv("ratings.tsv", sep="\t")
df
```

```python
X = df[["food", "ambience", "service"]]
y = df["rating"]

model = LinearRegression()
model.fit(X, y)
print(model.coef_)
```

A ranked list of coefficients: Food (0.7338), Service (0.3239) and Ambience (0.1338).

Based on the fitted model, food has the highest coefficient with 0.7338, meaning that it has the most influence on the overall rating. I would recommend the owner to fix their quality of food first to increase their overall rating and bring in more customers. Secondly, they could focus on the quality of service since it is ranked as second with coefficient of 0.3239. The ambience follows last with the coefficient 0.1338.
