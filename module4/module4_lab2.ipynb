```python
import pandas as pd
df = pd.read_csv("allbacks.csv")

df
```

Sort the values of "cover" column and call it a "sorted_series"

```python
sorted_series = df["cover"].sort_values()
```

```python
print(sorted_series)
```

Find the middle index by performing floor division and use that index to find the median and store it in "median_cover"

```python
median_cover = sorted_series.iloc[len(sorted_series) // 2]
```

```python
print(median_cover)
```

The median cover is pb (paperback), which happens to be the middle value in the sorted cover list. 

Find the mean weight of all books

```python
print(df["weight"].mean())
```

The mean weight of all books is about 706.67 grams.

Find the mean weight within each cover type (hardcover vs. paperback)

```python
df.loc[df["cover"] == "hb", "weight"].mean()
```

```python
df.loc[df["cover"] == "pb", "weight"].mean()
```

```python
df.groupby(by="cover", dropna=False).mean()
```

The mean weight for hardbacks is 796.429 grams. The mean weight for paperbacks is 628.125 grams. I can clearly see that the paperbacks weigh significantly less than the hardbacks, with paperbacks' mean weight weighing less than the overall mean weight for all books.
