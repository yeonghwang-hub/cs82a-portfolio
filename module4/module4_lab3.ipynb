```python
import pandas as pd
df = pd.read_csv("truck.csv")
df
```

```python
df.corr(numeric_only=True)
```

I see that the Pearson's r value between the list price and best price is about 0.99854. Its magnitude is very strong and positive. I can say that the list price is strongly associated with the best price.

```python
ax = df.plot.scatter(x="X", y="Y", title="List Price vs Best Price Distribution")
ax.set_xlabel("List Price in $1000s")
ax.set_ylabel("Best Price in $1000s")
```

The scatter shows a tightly hugged straight line that moves in the upward direction. This tells me that the scatter is showing positive strength.

Based on the Pearson's r value and the scatter, I predict that a list price of 22.0 would have a best price of 19.0. I am 99% confident that my prediction would be accurate based on how straight of a line the scatter shows. According to the scatter, the value at list price 22 (which is empty) would be around 19 (best price) in order for the value to fit into the line. The Pearson's r value tells me that the list price and best price are strongly correlated to one another, so such prediction can be made.
