```python
import pandas as pd
df = pd.read_csv("sales_clean.csv")

df
```

```python
print(f"Mean of price is {df["price"].mean()}")
print(f"Median of price is {df["price"].median()}")
print(f"Standard Deviation of price is {df["price"].std()}")
print(f"Mean of qty is {df["qty"].mean()}")
print(f"Median of qty is {df["qty"].median()}")
print(f"Standard Deviation of qty is {df["qty"].std()}")
```

To report the value for price in this case, I would use median. Because the data is skewed, extreme outliers could distort the data if I were to use mean. To minimize the impact of those outliers, I would use median.

```python
df["price"].plot(kind="hist", bins=20)
```

The histogram of "price" is right-skewed graph with a long tail to the right.

```python
df["revenue"] = df["price"] * df["qty"]
```

```python
df.plot.scatter(x="qty", y="revenue")
```

By looking at the scatter, the plots are rising, so I could say that the variables qty and revenue have a positive relationship. By looking at the Pearson's r value between qty and revenue which is 0.312793, I could say that its strength is moderate. I do need to be cautious of the two plots under qty valued at 5 where I see two extremely high revenues. They look like outliers that could distort the data.
