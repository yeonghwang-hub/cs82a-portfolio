Based on my experience of moving from local Jupyter environment to Google Colab, there were things that differed and some similarities between them as well. First of all, when I first installed Jupyter and Anaconda on my laptop, there were several installation process that I needed to carefully follow. I don't recall doing one on Google Colab! Also, every time I need to use Jupyter, I would use terminal and activate it from there, but for Google Colab, I can just simply go to the google colab website and open a new notebook from there like opening a new Google Doc. The similarities between them were in running the program in general, like Shift+Enter command and Run All command. For a possible risk, I personally thought that working on Jupyter felt a little more secure than working on Google Colab. Maybe it has to do with the several tasks before setting up the Jupyter work environment, but I felt like working on Google Colab was like working on Google Docs, easy and straight-forward, but a little less private.
