```python
import pandas as pd
df = pd.read_csv("usarrests.csv")
df
```

```python
df.dtypes
```

By looking at the table, I can see that the first column is named "Unnamed: 0" while it could be named as "State" or "State Name". I decided to rename the column as "State".

```python
df = df.rename(columns={"Unnamed: 0" : "State"})
df.shape
```

```python
df.isna().sum()
```

By running the .isna() and .sum() functions to count the number of na in the table, I was able to count 1 missing data for Georgia's Assault value. I decided to drop the row for Georgia because the assault rate can't possibly be 0 as a typo when its murder rate is 17.4.

```python
df = df.drop(index=9)
df.shape
```

```python
df.isna().sum()
```

```python
df.describe()
```

Based on the describe() function, I see that New York has the minimum urban population with only 6%. This makes zero sense and almost impossible especially when it's compared to other states, so I decided to drop it from the data.

```python
df = df.drop(index=31)
df.shape
```

```python
df
```

By looking at the data, I see Iowa's urban population is 570% which is purely impossible. I'm going to treat it as a typo and change it to 57%.

```python
a = df["UrbanPop"][14]
b = 57
df = df.replace(a, b)
df
```

I see that South Carolina's Assault rate at 879.0 seems a little too extreme. I am going to drop the row for South Carolina.

```python
df = df.drop(index=39)
df.shape
```

```python
df
```

```python
df.describe()
```
