```python
import pandas as pd
df = pd.read_csv("bridges.csv")
```

I'm going to make an assumption that bridges cannot have the length with value NaN or na. I will drop the bridges with length NaN.

```python
df
```

```python
df.isna().sum()
```

```python
df[df['length'].isna()]
```

I see that a few bridges have length value NaN already, like ones labeled E1, E3, E6 and so on. By using df[df['length'].isna()] I can visually see all the bridges with length NaN. By using isna() and sum() functions, I see a total of 27 bridges have length with value NaN. I will drop them and make a clean dataframe that only contains bridges with lengths that are in number values.

```python
df_cleaned = df.dropna(subset=['length'])
```
